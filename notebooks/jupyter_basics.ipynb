## 1.  Two kinds of cell



A notebook is a column of **cells**.  There are two kinds.

**Text cells**, like this one, hold prose.  Double-click one to edit it, and
run it to set it back to formatted text.

**Code cells** hold Python.  Running one sends the code to be executed and
prints whatever comes back underneath.  The next cell is a code cell.



```python
%xmode Plain

print("Hello from a code cell")
```

Ignore that first line for now; section 6 comes back to it.



## 1.  Running a cell



Click on a cell to select it, then either press the ▶ button in the toolbar
or — much faster, and worth learning now — hold **Shift** and press
**Return**.

Shift+Return runs the cell and moves to the next one, so you can walk down a
notebook by holding Shift and tapping Return.  Control+Return runs the cell
and stays put, which is what you want while fixing one line repeatedly.

A cell that produces a value shows it without `print`, but only for the
*last* line:



```python
2 + 2
40 + 2
```

Only `42` appears.  The first line was computed and thrown away.  If you want
both, use `print` for both.



## 1.  The number in the brackets, and why order matters



To the left of a code cell you will see `[ ]`.  After you run it, a number
appears: `[1]`, `[2]`, and so on.  **That number is the order in which cells
were run — not the order they appear on the page.**  A cell running right
now shows `[*]`.

This is the single most important thing in this tutorial, because a notebook
does not care what order you run things in, and it will not warn you.



```python
n = 0
```

```python
n = n + 1
print("n is now", n)
```

Run that second cell again.  And again.  `n` keeps climbing, because the cell
does not reset anything — it adds one to whatever `n` happens to be.  The
notebook you are looking at is unchanged, but the answer is different.

Two habits follow, and they are worth adopting immediately.

-   **Work downwards:** run cells in the order they appear, top to bottom.
-   **When confused, start clean:** *Kernel > Restart Kernel and Run All Cells*.
    If the notebook is correct, this always works.  If it does not, the
    notebook depends on something you did out of order, and you have just
    found a real bug rather than a mystery.



## 1.  The kernel



Behind the notebook is a **kernel**: a Python process that holds every variable
you have defined and runs the code you send it.  The notebook is the page;
the kernel is the memory.

At the top right you can see its state.  An empty circle means idle and
ready; a filled circle means it is busy.  A cell showing `[*]` and a filled
circle means your code is still running — wait rather than clicking again.

Restarting the kernel throws away every variable, which is exactly what you
want when things have got into a tangle.  Nothing in your notebook is lost;
you simply have to run the cells again.

This server offers two kernels.  **Python 3** is the one to use, and the one
these tutorials and the session notebooks expect.  There is also **Python 3
(LSMS development)**, which exists for one optional exercise later in the
workshop.  If you ever see an error saying a module cannot be found, check
which kernel you are on before anything else.



## 1.  Asking the notebook instead of asking a person



Three keystrokes will answer most of the questions you would otherwise have
to look up.  None of them can be shown by running a cell — they are things
you press — so try each one now.



### Tab completes names



Type `pd.DataFrame.` into an empty cell and, without running it, press **Tab**.
A list appears of everything a DataFrame can do.  Keep typing to narrow it:
`pd.DataFrame.grou` then Tab leaves very little doubt.

This is also how you check whether something exists, which matters here: if
the name you half-remember does not appear in the list, it is not there, and
no amount of running the cell will make it be.



```python
import pandas as pd

# Run this cell first so `pd' exists, then try Tab in a cell of your own.
print("pandas is loaded; now try  pd.DataFrame.<Tab>  in an empty cell")
```

### Shift+Tab shows how to call something



Put your cursor inside the brackets of a function you are about to call —
`pd.read_csv(` — and press **Shift+Tab**.  A panel appears giving the
arguments it takes and what they mean.  Press it twice for the full
documentation.



### A question mark prints the manual



Appending `?` to a name and running the cell prints its documentation:



```python
# The same thing, in a form that works in any Python:
help(len)
```

In a notebook you can write `len?` instead, which is shorter and opens a
panel at the bottom of the window.



## 1.  Magics



Lines beginning with `%` are not Python.  They are **magics**: instructions to
the notebook itself rather than code to be run.  You have already run one:
the first code cell of this notebook began with

    %xmode Plain

as does every other notebook in the workshop.  It shortens error messages to the line that failed and why, instead of
forty lines of library internals.

Two more worth knowing.  `%whos` lists everything currently defined, which is
the quickest way to see what the kernel is actually holding:



```python
%whos
```

And `%explain`, which is particular to this workshop.  It sends your question
— or, with nothing after it, the error you just got — to an assistant that
has been given the `lsms_library` documentation, and prints a short answer.



```python
%explain what does the percent sign mean at the start of a line in Jupyter?
```

The more useful form takes no argument at all.  The next time a cell fails,
run `%explain` in the cell below it and it will read the error for you.

Run it in the cell *below* the one you are asking about: it can see the cell
you just ran, what that cell returned — the shape of a table and the names
of its columns, never the data in it — and the names you have defined.  So
a question about a column in front of you needs no preamble:

    ll.Country('Uganda').household_roster()

    %explain what does Affinity mean in this context?

Three things to know about it.  It is a helper, not an oracle: it is good on
Python and on this library, and it will decline questions about which poverty
line you should use, because those are for the workshop.  It has been told
not to guess — if it says a method does not exist, believe it.  And if it
is unavailable it says so in one line and everything else keeps working.



## When something goes wrong



In this order, and it is rarely necessary to go past the second step.

1.  **Read the last line of the error.**  It names the kind of problem and
    usually the value that caused it.  The notebooks are set up to show you
    your own failing line and little else.
2.  **Kernel > Restart Kernel and Run All Cells.**  This fixes anything caused
    by running cells out of order, which is most things.
3.  **Run `%explain` in the next cell.**
4.  **Ask.**  Errors in this workshop are information about the materials, not
    evidence about you — and a question asked aloud usually turns out to be
    three people's question.
